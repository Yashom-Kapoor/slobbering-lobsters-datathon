# Clean dataset

### Load OECD-Dataset into a CSV

```python
import pandas as pd 

df = pd.read_csv("../../data/OECD-Dataset.csv")
df.head()
```

### Clean years
Some years are strings and others are integers, turn them all to integers.

```python
df.groupby('year').count()
```

```python
df_2020_23 = df[df['year'] == '2020-2023']
df_2020_23
```

Handle ranges of 2020-2023: Split into 2020, 2021, 2022, and 2023 with funding evenly distributed across 4 years.

```python
def handle_range(row):
    if (row['year'] == '2020-2023'):
        row['year'] = [2020, 2021, 2022, 2023]
        row['usd_disbursements_defl'] = row['usd_disbursements_defl'] / 4
    elif (type(row['year']) is str):
        row['year'] = int(row['year'])
    return row

df = df.apply(handle_range, axis=1)
df = df.explode('year')
df.head()
```

```python
df.groupby('year').count()
```

### Explore organization_name, region, country, grant recipient

```python
df.groupby('organization_name').count()
```

```python
df.groupby('region').count()
```

Handle region: split on ; and distribute evenly

```python
def handle_regions(row):
    if (';' in row['region']):
        arr = row['region'].split(';')
        arr_strip = [a.strip() for a in arr]
        row['region'] = arr_strip
        row['usd_disbursements_defl'] = row['usd_disbursements_defl'] / (len(arr_strip))
    else:
        row['region'] = row['region'].strip()
    return row

df = df.apply(handle_regions, axis=1)
df = df.explode('region')
df.groupby('region').count()
```
