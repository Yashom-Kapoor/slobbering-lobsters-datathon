# View funding by region, country, and year=

```python
import pandas as pd 
import plotly.express as px

df = pd.read_csv("../../data/clean_year_region.csv")
df.head()
```

```python
df[["country", "usd_disbursements_defl"]][df["country"].isin(["India", "Ukraine", "Afghanistan", "Kenya"])].groupby("country").sum()
```

```python
test = df[["country", "usd_disbursements_defl"]][df["country"].isin(["India", "Ukraine", "Afghanistan", "Kenya"])].groupby("country", as_index=False).sum()
fig = px.choropleth(
    test,
    locations="country",
    locationmode="country names",
    color="usd_disbursements_defl"
)

fig.show()
```
